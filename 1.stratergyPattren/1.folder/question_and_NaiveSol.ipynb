# Question
- Design a Payment Processing System

You are building an e-commerce application. Customers should be able to pay using different payment methods.

The system must support:
1. UPI payment
2. Credit Card payment
3. PayPal payment
Requirements:
- The customer selects a payment method.
- The system processes the payment using the selected method.
- Each payment method has its own payment-processing logic.
- New payment methods should be addable in the future.
- The main payment system should handle the payment request.

## Naive Solution

```python
class Payment:
    def process_payment(self, payment_type, amount):

        if payment_type == "upi":
            print(f"Processing UPI payment of {amount}")

        elif payment_type == "credit_card":
            print(f"Processing Credit Card payment of {amount}")

        elif payment_type == "paypal":
            print(f"Processing PayPal payment of {amount}")

        else:
            print("Invalid payment method")


payment = Payment()
payment.process_payment("upi", 1000)
```

### Issues
-  Violates Open/Closed Principle (OCP)
Every time we introduce a new payment method, we modify the existing process_payment() method by adding another if-elif block.
OCP says:Software entities should be open for extension but closed for modification.


-  Violates Single Responsibility Principle (SRP)
The Payment class handles:
 Selecting the payment method.
 Processing UPI.
 Processing credit cards.
 Processing PayPal.
It has multiple reasons to change.


- Tight coupling
The payment class knows the internal implementation of every payment method.
