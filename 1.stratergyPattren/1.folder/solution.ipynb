```python
from abc import ABC, abstractmethod

class PaymentStrategy(ABC):
    @abstractmethod
    def process_payment(self, amount):
        pass
```

```python
class UPIPayment(PaymentStrategy):
    def process_payment(self, amount):
        print(f"Processing UPI: {amount}")


class CreditCardPayment(PaymentStrategy):

    def process_payment(self, amount):
        print(f"Processing Credit Card: {amount}")


class PayPalPayment(PaymentStrategy):

    def process_payment(self, amount):
        print(f"Processing PayPal: {amount}")
```

```python
class Payment:
    def __init__(self, strategy: PaymentStrategy):
        self.strategy = strategy
    def make_payment(self, amount):
        self.strategy.process_payment(amount)
    def change_strategy(self,strategy:PaymentStrategy):
        self.strategy = strategy
```

```python
payment = Payment(UPIPayment())
payment.make_payment(1000)
payment.change_strategy(CreditCardPayment())
payment.make_payment(2000)
```
