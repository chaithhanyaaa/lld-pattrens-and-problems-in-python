 # Question
 You're building a backend notification service.

The application needs to send notifications through different channels:
1. Email notification
2. SMS notification
3. WhatsApp notification
Requirements:
- The user selects a notification type.
- The system creates the appropriate notification object.
- Each notification type has its own sending logic.
- The application should support adding new notification types in the future.

# Simple Factory

```python
from abc import ABC ,abstractmethod
class Notification(ABC):
  @abstractmethod
  def send_notification(self, message):
    pass

class EmailNotification(Notification):
  def send_notification(self, message):
    print(f"Sending email notification: {message}")

class SMSNotification(Notification):
  def send_notification(self, message):
    print(f"Sending SMS notification: {message}") 

class WhatsappNotification(Notification):
  def send_notification(self, message):
    print(f"Sending WhatsApp notification: {message}")  

class NotificationFactory:
  @staticmethod
  def create_notification(notification_type):
    if notification_type == "email":
      return EmailNotification()
    elif notification_type == "sms":
      return SMSNotification()
    elif notification_type == "whatsapp":
      return WhatsappNotification()
    else:
      raise ValueError("Invalid notification type")

notification = NotificationFactory.create_notification("email")
notification.send_notification("Hello, this is an email notification!")
```

## GOF Solution- Gang of four

```python
from abc import ABC, abstractmethod


# Product interface
class Notification(ABC):

    @abstractmethod
    def send(self):
        pass


# Concrete products
class EmailNotification(Notification):
    def send(self):
        print("Sending Email")


class SMSNotification(Notification):
    def send(self):
        print("Sending SMS")


# Creator
class NotificationCreator(ABC):

    @abstractmethod
    def create_notification(self):
        pass

    def notify(self):
        notification = self.create_notification()
        notification.send()


# Concrete creators
class EmailCreator(NotificationCreator):

    def create_notification(self):
        return EmailNotification()


class SMSCreator(NotificationCreator):

    def create_notification(self):
        return SMSNotification()


# Client
creator = EmailCreator()
creator.notify()
creator = SMSCreator()
creator.notify()
```

# dictonary based

```python
from abc import ABC ,abstractmethod

class Notification(ABC):
  @abstractmethod
  def send_notification(self, message):
    pass


class EmailNotification(Notification):
  def send_notification(self, message):
    print(f"Sending email notification: {message}")

class SMSNotification(Notification):
  def send_notification(self, message):
    print(f"Sending SMS notification: {message}") 

class WhatsappNotification(Notification):
  def send_notification(self, message):
    print(f"Sending WhatsApp notification: {message}")  

    


class NotificationFactory:
  registry={
    "email":EmailNotification, # we stored class not object
    "sms":SMSNotification,
    "whatsapp":WhatsappNotification
  }

  @classmethod
  def add_notification_type(cls,notification_type,notification_class):
    cls.registry[notification_type]=notification_class

  @classmethod
  def create(cls,notification_type):
    if notification_type in cls.registry:
      return cls.registry[notification_type]()
    else:
      raise ValueError(f"Notification type '{notification_type}' is not supported.")
```

```python
notification = NotificationFactory.create("email")
notification.send("Order shipped")
```

- dictonary lookup
- no inheritance
- number of classes less
- less complexity
