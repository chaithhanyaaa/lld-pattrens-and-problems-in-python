# Question
LLD Problem: Cross-Platform UI Component System
Design a UI component system that supports multiple operating systems.

Supported operating systems:
- Windows
- macOS
Each operating system has:
- Button
- Checkbox
Requirements
1. Create UI components based on the selected operating system.
2. If Windows is selected, create WindowsButton and WindowsCheckbox.
3. If macOS is selected, create MacButton and MacCheckbox.
4. Adding another operating system later should be manageable.
5. Client code should not directly depend on concrete UI classes.

# Naive

```python
class WindowsButton:
    def render(self):
        print("Rendering Windows Button")


class WindowsCheckbox:
    def render(self):
        print("Rendering Windows Checkbox")


class MacButton:
    def render(self):
        print("Rendering Mac Button")


class MacCheckbox:
    def render(self):
        print("Rendering Mac Checkbox")


class UI:
    def __init__(self, os_type):
        if os_type == "windows":
            self.button = WindowsButton()
            self.checkbox = WindowsCheckbox()

        elif os_type == "mac":
            self.button = MacButton()
            self.checkbox = MacCheckbox()

        else:
            raise ValueError("Unsupported OS")

    def render(self):
        self.button.render()
        self.checkbox.render()


ui = UI("windows")
ui.render()
```

Problems with the naive implementation
1. Open/Closed Principle violation
If we add Linux, we must modify the UI class by adding another elif condition.

2. Tight coupling
The UI class directly creates concrete classes:
- WindowsButton
- WindowsCheckbox
- MacButton
- MacCheckbox
It knows the implementation details of every operating system.

3. Consistency problem
Nothing prevents accidentally combining a WindowsButton with a MacCheckbox.

4. Poor scalability
As operating systems and UI components increase, the conditional logic inside UI becomes increasingly complicated.

# Pattren

```python

from abc import ABC, abstractmethod


# 1. Abstract Products
class Button(ABC):
    @abstractmethod
    def render(self):
        pass


class Checkbox(ABC):
    @abstractmethod
    def render(self):
        pass


# 2. Concrete Products - Windows
class WindowsButton(Button):
    def render(self):
        print("Rendering Windows Button")


class WindowsCheckbox(Checkbox):
    def render(self):
        print("Rendering Windows Checkbox")


# 3. Concrete Products - Mac
class MacButton(Button):
    def render(self):
        print("Rendering Mac Button")


class MacCheckbox(Checkbox):
    def render(self):
        print("Rendering Mac Checkbox")


# 4. Concrete Products - Linux
class LinuxButton(Button):
    def render(self):
        print("Rendering Linux Button")


class LinuxCheckbox(Checkbox):
    def render(self):
        print("Rendering Linux Checkbox")


# 5. Abstract Factory
class Factory(ABC):
    @abstractmethod
    def create_button(self):
        pass

    @abstractmethod
    def create_checkbox(self):
        pass


# 6. Concrete Factories
class WindowsFactory(Factory):
    def create_button(self):
        return WindowsButton()

    def create_checkbox(self):
        return WindowsCheckbox()


class MacFactory(Factory):
    def create_button(self):
        return MacButton()

    def create_checkbox(self):
        return MacCheckbox()


class LinuxFactory(Factory):
    def create_button(self):
        return LinuxButton()

    def create_checkbox(self):
        return LinuxCheckbox()


# 7. Factory Registry
class FactoryProvider:
    registry = {
        "windows": WindowsFactory,
        "mac": MacFactory,
        "linux": LinuxFactory
    }

    @classmethod
    def get_factory(cls, os_type):
        factory_class = cls.registry.get(os_type.lower())

        if factory_class is None:
            raise ValueError("Unsupported operating system")

        return factory_class()


# 8. Client
class UI:
    def __init__(self, factory: Factory):
        self.button = factory.create_button()
        self.checkbox = factory.create_checkbox()

    def render(self):
        self.button.render()
        self.checkbox.render()


# 9. Application
os_type = "windows"

factory = FactoryProvider.get_factory(os_type)

ui = UI(factory)
ui.render()
```
