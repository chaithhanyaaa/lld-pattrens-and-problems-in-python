You are designing a Payment system.
There are payment methods:

UPI
Card

Each payment has:

pay() method

👉 BUT there is a new rule:

There are two regions:

INDIA

US

Rules:

In INDIA → use IndianUPI, IndianCard

In US → use USUPI, USCard

⚠️ Important:

You must NOT mix regions
(e.g., IndianUPI + USCard ❌)

```python
from abc import ABC,abstractmethod

class upi(ABC):
  @abstractmethod
  def pay(self, amount):
    pass

class card(ABC):
  @abstractmethod
  def pay(self, amount):
    pass

class indianCard(card):
  def pay(self, amount):
    print(f"paying {amount} using indian card")

class indianUpi(upi):
  def pay(self, amount):
    print(f"paying {amount} using indian upi")


class UsCard(card):
  def pay(self, amount):
    print(f"paying {amount} using us card")

class UsUpi(upi):
  def pay(self, amount):
    print(f"paying {amount} using us upi")



class Factory(ABC):
  @abstractmethod
  def createCard(self):
    pass
  
  @abstractmethod
  def createUpi(self):
    pass

class indianFactory(Factory):
  def createCard(self):
    return indianCard()
  def createUpi(self):
    return indianUpi()


class UsFactory(Factory):
  def createCard(self):
    return UsCard()
  def createUpi(self):
    return UsUpi()


class FactoryProvider:
  registry={
    "us":UsFactory,
    "india":indianFactory
  }

  @classmethod
  def get_factory(cls, country):
      factory_class = cls.registry.get(country.lower())

      if factory_class is None:
          raise ValueError("Select valid country")

      return factory_class()
    


class Payment:
  def __init__(self,factory):
    self.factory=factory
    self.upi=factory.createUpi()
    self.card=factory.createCard() 
  def payUpi(self,amount):
    self.upi.pay(amount)
  def payCard(self,amount):
    self.card.pay(amount)

factory=FactoryProvider.get_factory("india")
payment=Payment(factory)
payment.payUpi(500)

    

    


```
