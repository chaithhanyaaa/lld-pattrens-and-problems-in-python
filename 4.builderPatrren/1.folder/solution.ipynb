# Question
Design a Computer Configuration System.

A computer can have:
- CPU
- RAM
- Storage
- GPU
- Operating System
Requirements:
1. Users should be able to configure different computers.
2. Some components are optional (GPU, for example).
3. Different configurations should be supported (Gaming PC, Office PC, Basic PC).
4. Object creation should be readable and manageable.
5. Avoid constructors with too many parameters.

# Naive Solution


```python
class Computer:

    def __init__(self, cpu, ram, storage, os):
        self.cpu = cpu
        self.ram = ram
        self.storage = storage
        self.os = os
        self.gpu = None

    def add_gpu(self, gpu):
        self.gpu = gpu


computer = Computer(
    cpu="Intel i7",
    ram="16GB",
    storage="1TB SSD",
    os="Windows"
)

computer.add_gpu("NVIDIA RTX 4060")
```

1. Too many configuration methods: With 15 optional components, the Computer class may need 15 separate methods.

2. Incomplete objects: Someone might create a computer but forget to configure required components.

3. Repeated configuration: Gaming PC, Office PC and Basic PC configurations may require repeating the same setup logic.

4. Poor readability: When configuring many components, the creation process becomes scattered across multiple method calls.

5. Validation becomes difficult: We need to ensure the computer is properly configured before using it.

# Solution


```python

# 1. Product
class Computer:

    def __init__(self, object):

        self.cpu = object.cpu
        self.ram = object.ram
        self.storage = object.storage
        self.os = object.os
        self.gpu = object.gpu
        self.wifi = object.wifi
        self.bluetooth = object.bluetooth

    def show_configuration(self):
        print("\nComputer Configuration")
        print("----------------------")
        print("CPU:", self.cpu)
        print("RAM:", self.ram)
        print("Storage:", self.storage)
        print("OS:", self.os)
        print("GPU:", self.gpu)
        print("WiFi:", self.wifi)
        print("Bluetooth:", self.bluetooth)


# 2. Builder
class ComputerBuilder:

    def __init__(self):
        self.cpu = None
        self.ram = None
        self.storage = None
        self.os = None
        self.gpu = None
        self.wifi = None
        self.bluetooth = None

    def set_cpu(self, cpu):
        self.cpu = cpu
        return self

    def set_ram(self, ram):
        self.ram = ram
        return self

    def set_storage(self, storage):
        self.storage = storage
        return self

    def set_os(self, os):
        self.os = os
        return self

    def set_gpu(self, gpu):
        self.gpu = gpu
        return self

    def set_wifi(self, wifi):
        self.wifi = wifi
        return self

    def set_bluetooth(self, bluetooth):
        self.bluetooth = bluetooth
        return self

    def build(self):
        # Validate mandatory fields
        if not all([self.cpu, self.ram, self.storage, self.os]):
            raise ValueError("Missing mandatory components")

        # Create final product
        return Computer(self)


# 3. Director (optional)
class ComputerDirector:

    @staticmethod
    def create_gaming_pc(builder):

        return (
            builder
            .set_cpu("Intel i9")
            .set_ram("64GB")
            .set_storage("2TB SSD")
            .set_gpu("NVIDIA RTX 5090")
            .set_wifi("WiFi 7")
            .set_bluetooth("Bluetooth 5.4")
            .set_os("Windows")
            .build()
        )

    @staticmethod
    def create_office_pc(builder):

        return (
            builder
            .set_cpu("Intel i5")
            .set_ram("16GB")
            .set_storage("512GB SSD")
            .set_wifi("WiFi 6")
            .set_os("Linux")
            .build()
        )


# 4. Client

# Custom Gaming PC
gaming_pc = (
    ComputerBuilder()
    .set_cpu("AMD Ryzen 9")
    .set_ram("32GB")
    .set_storage("1TB SSD")
    .set_gpu("RTX 4080")
    .set_os("Windows")
    .build()
)

gaming_pc.show_configuration()


# Custom Office PC
office_pc = (
    ComputerBuilder()
    .set_cpu("Intel i5")
    .set_ram("16GB")
    .set_storage("512GB SSD")
    .set_os("Ubuntu")
    .build()
)

office_pc.show_configuration()


# Predefined Gaming PC using Director
director = ComputerDirector()

predefined_gaming = director.create_gaming_pc(
    ComputerBuilder()
)

predefined_gaming.show_configuration()


# Example of validation failure
try:
    invalid_pc = (
        ComputerBuilder()
        .set_cpu("Intel i5")
        .set_ram("8GB")
        .build()
    )
except ValueError as error:
    print("\nError:", error)
```
