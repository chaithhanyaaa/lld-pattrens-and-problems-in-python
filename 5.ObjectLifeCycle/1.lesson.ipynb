# 2.Object refernce and counting

```python
class Student:
    def __init__(self, name):
        self.name = name

```

```python
import sys

s1 = Student("Chaithanya")
print(sys.getrefcount(s1))

s2 = s1
print(sys.getrefcount(s1))

s3 = s1
print(sys.getrefcount(s1))

del s1
print(sys.getrefcount(s2))
s2.name
# del removes a reference (the variable binding), not necessarily the object itself.
```

# 3.Garabage Collection

```python
# circular reference example
class A:
    pass
class B:
    pass
a=A()
b=B()
# reference count of a,b is 1
a.other=b
b.other=a
# now refernce count of a,b is 2
del a
del b
# now refernces are deleted,but A object is still pointing to B and B object pointing to A
# reference count of A and B is 1, so they are not garbage collected
```

- Python's cyclic garbage collector can detect that these objects are unreachable, even though they reference each other.
