##### Process
- process is an instance of a running process
ex:Chrome,VsCode,spotify
- oS creates these process to run these application.
- Each has its own memory,resources,execution state,pid 

```python
import os

print("Process ID:", os.getpid())
```

- Your operating system schedules CPU time for these processes.when we have multiple.
- On a multicore CPU, different processes can execute simultaneously on different cores.
- Concurrency: Multiple tasks make progress during overlapping periods.
- Parallelism: Multiple tasks literally execute at the same time.

#### Thread
- A thread is the smallest unit of execution that the operating system schedules within a process.
- share headmemory,global varibles,resources such as file descriptors,each thread has its own stack,registers,instruction pointer.

```python
import threading
def task():
    print("Thread ID:", threading.get_ident())

# Create and start threads
t1 = threading.Thread(target=task)
t2 = threading.Thread(target=task)

t1.start()
t2.start()

t1.join()
t2.join()
print("Main Thread ID:", threading.get_ident())
```

# Multithreading

```python
import threading
import time

def task(name):
    print(f"{name} started")
    time.sleep(2)
    print(f"{name} completed")

t1 = threading.Thread(target=task, args=("Task 1",))
t2 = threading.Thread(target=task, args=("Task 2",))

t1.start()
t2.start()

t1.join()
t2.join()

print("All tasks completed")
```

# Multiprocessing

```python
import multiprocessing
import time

def task(name):
    print(f"{name} started")
    time.sleep(2)
    print(f"{name} completed")

if __name__ == "__main__":
    p1 = multiprocessing.Process(target=task, args=("Task 1",))
    p2 = multiprocessing.Process(target=task, args=("Task 2",))

    p1.start()
    p2.start()

    p1.join()
    p2.join()

    print("All tasks completed")
```

```python
import threading

counter = 0

def task():
    global counter
    counter += 1
    print("Inside thread:", counter)

t1 = threading.Thread(target=task)
t2 = threading.Thread(target=task)

t1.start()
t2.start()

t1.join()
t2.join()

print("Main thread:", counter)
```

```python
import multiprocessing

counter = 0

def task():
    global counter
    counter += 1
    print("Inside process:", counter)

if __name__ == "__main__":
    p1 = multiprocessing.Process(target=task)
    p2 = multiprocessing.Process(target=task)

    p1.start()
    p2.start()

    p1.join()
    p2.join()

    print("Main process:", counter)
```
